```python
import json, torch
from torch.utils.data import DataLoader
from segres import SolarFilamentDataset, build_model, get_train_transform, get_val_transform, DiceBCELoss, set_seed, train_one_epoch, validate

cfg = json.load(open('config.json'))
set_seed(cfg['seed'])
device = torch.device('cpu')

common = dict(images_dir=cfg['train_images_dir'], 
annotations_json=cfg['train_annotations'],
            tile_size=128, 
overlap=cfg['overlap'])
train_ds = SolarFilamentDataset(['20110109104734Ch.jpeg', 
'20110114105034Ch.jpeg'], transform=get_train_transform(), 
**common)
val_ds = SolarFilamentDataset(['20110119082634Lh.jpeg'], 
transform=get_val_transform(), **common)

train_loader = DataLoader(train_ds, batch_size=4)
val_loader = DataLoader(val_ds, batch_size=4, shuffle=False,
num_workers=0)

model = build_model().to(device)
loss_fn = DiceBCELoss()
opt = torch.optim.Adam(model.parameters(), lr=1e-4)

print(train_one_epoch(model, train_loader, opt, loss_fn, device))
print(validate(model, val_loader, loss_fn, device))
```
